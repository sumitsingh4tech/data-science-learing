```python
from ollama import Client
client = Client(
  host='http://18.232.128.120:11434',
)
response = client.chat(model='llama3', messages=[
  {
    'role': 'user',
    'content': 'Why is the sky blue?',
  },
], stream = True)

for chunk in response:
    print(chunk["message"]["content"], end="", flush=True)
```
